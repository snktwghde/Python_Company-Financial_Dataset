```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("F:/IT Vedant_Python/Financials.csv")
df
```

```python
#check data types of the columns
df.info()
```

```python
#checking null values
df.isnull().sum()
```

```python
#check column exact names
df.columns.tolist()
```

```python
#removing spaces
df.columns = df.columns.str.strip()
```

```python
df.columns.tolist()
```

```python
#replacing values
df['Units Sold'] = df['Units Sold'].replace({'\$': '', ',': ''}, regex=True)
df['Manufacturing Price'] = df['Manufacturing Price'].replace({'\$': '', ',': ''}, regex=True)
df['Sale Price'] = df['Sale Price'].replace({'\$': '', ',': ''}, regex=True)
df['Gross Sales'] = df['Gross Sales'].replace({'\$': '', ',': ''}, regex=True)
df['Sales'] = df['Sales'].replace({'\$': '', ',': ''}, regex=True)
df['COGS'] = df['COGS'].replace({'\$': '', ',': ''}, regex=True)
df['Profit'] = df['Profit'].replace({'\$': '', ',': ''}, regex=True)
df['Discounts'] = df['Discounts'].replace({'\$': '', ',': '','-':''}, regex=True)
df
```

```python
#replacing blank values with 0
df = df.replace(r'^\s*$', 0, regex=True)  # Replace blank or space-only strings with 0
df = df.fillna(0)
df
```

```python
df.info()
```

```python
df['Discounts'] = pd.to_numeric(df['Discounts'], errors='coerce')
```

```python
df['Profit'] = pd.to_numeric(df['Profit'], errors='coerce')
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
```

```python
#unique products
df['Product'].unique()
```

```python
#unique segments
df['Segment'].unique()
```

```python
#max sales
df['Sales'].max()
```

```python
#min sales
df['Sales'].min()
```

```python
#How has the segment’s Revenue changed over time?
plt.figure(figsize=(10,6))
sns.lineplot(x='Year', y='Sales', hue='Segment', data=df)
plt.title("Sales Trend per Company Over Time")
plt.show()
```

```python
#How do Profits for each product compare across time?

sns.barplot(x='Product',y='Profit',hue='Year',data=df,palette='deep')
```

```python
#Which segments product contribute the most revenue?
sns.barplot(x='Product',y='Sales',hue='Segment',data=df)
```

```python
#Is there a strong correlation between Sales and Profit?
sns.scatterplot(x='Sales',y='Profit',hue='Year',data=df)
```

```python
#How do higher discounts affect sales or profit margins for each product?
plt.figure(figsize=(20,10))
sns.scatterplot(x='Discounts',y='Profit',hue='Product',data=df,palette="bright")
plt.show()
```

```python
#Which months or quarters bring the highest revenue?
sns.barplot(x='Sales',y='Month Name',hue='Year',data=df,palette='pastel')
```
